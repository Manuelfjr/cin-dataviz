# Imports

```python
from nb_utils import set_root
PROJECT_DIR = set_root(2)
```

```python
import pandas as pd
import plotly.express as px
import hiplot as hip
import plotly.graph_objects as go
```

# Params

```python
path_data = PROJECT_DIR / "data"
path_intermediate = path_data / "02_intermediate"

file_path_corr = path_intermediate / "data_horm_concat_corr.parquet"
file_path_all = path_intermediate / "data_horm_concat.parquet"
```

# Read

```python
data_input = pd.read_parquet(file_path_corr)
```

# Graphs

```python
# Cria uma paleta de cores baseada no hexadecimal
custom_colorscale = ["#FF4B3E", "#1C2127"]  # Lista com pelo menos duas cores

corr_matrix_feats = data_input.corr()
# Cria o heatmap
fig = px.imshow(
    corr_matrix_feats,
    labels=dict(color="Correlation"),
    y=corr_matrix_feats.columns,
    x=corr_matrix_feats.index,  # Nomes das colunas como índices
    color_continuous_scale=custom_colorscale  # Define o esquema de cores personalizado
)

fig.update_xaxes(showticklabels=False)
fig.update_layout(height=1200) 

# Exibe o gráfico
fig.show()
```

```python
fig = px.scatter_matrix(data_input, color="Sperm C20:1 n-9", height=1800, color_continuous_scale=custom_colorscale)
fig.show()
```

```python
hiplot = hip.Experiment.from_dataframe(data_input)
hiplot.display()
```
